# Data Exploration

Quick look at the custom 3-class image dataset: counts per class, sample images, and image-size distribution.

```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
import cv2
import numpy as np

import sys; sys.path.append('..')
from src.config import TRAIN_DIR, VAL_DIR, TEST_DIR, CLASS_NAMES
```

## 1. Class distribution

```python
def count_per_class(folder):
    return {c: len(list((folder / c).glob('*'))) for c in CLASS_NAMES if (folder / c).exists()}

for split, folder in [('train', TRAIN_DIR), ('val', VAL_DIR), ('test', TEST_DIR)]:
    print(f'{split}: {count_per_class(folder)}')
```

## 2. Sample images per class

```python
fig, axes = plt.subplots(len(CLASS_NAMES), 4, figsize=(12, 3*len(CLASS_NAMES)))
for i, cls in enumerate(CLASS_NAMES):
    sample_paths = list((TRAIN_DIR / cls).glob('*'))[:4]
    for j, p in enumerate(sample_paths):
        img = cv2.imread(str(p))
        if img is None: continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        axes[i, j].imshow(img)
        axes[i, j].axis('off')
        if j == 0: axes[i, j].set_ylabel(cls, fontsize=12)
plt.tight_layout()
plt.show()
```

## 3. Image size distribution

```python
sizes = []
for cls in CLASS_NAMES:
    for p in (TRAIN_DIR / cls).glob('*'):
        img = cv2.imread(str(p))
        if img is not None: sizes.append(img.shape[:2])

sizes = np.array(sizes)
print(f'Mean H x W: {sizes.mean(0)}')
print(f'Min H x W:  {sizes.min(0)}')
print(f'Max H x W:  {sizes.max(0)}')
```
